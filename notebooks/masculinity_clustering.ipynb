```python
from pathlib import Path
import pandas as pd

pd.set_option('display.max_columns', None)
data_dir = Path('../data')

df = pd.read_csv(data_dir / 'masculinity.csv')

print(f"Jumlah responden : {df.shape[0]} baris")
print(f"Jumlah kolom     : {df.shape[1]} kolom")
print("\nDaftar seluruh kolom:")
print(list(df.columns))

print("\nContoh variasi jawaban pada beberapa kolom pertanyaan:")
for col in ['q0001', 'q0002', 'q0007_0001']:
    if col in df.columns:
        print(f"- {col}: {df[col].dropna().unique().tolist()}")

df.head(3)
```

```python
import numpy as np

# 1. Daftar 11 kolom pertanyaan frekuensi perilaku (q0007_0001 - q0007_0011)
q7_cols = [f'q0007_00{i:02d}' for i in range(1, 12) if f'q0007_00{i:02d}' in df.columns]

# 2. Peta konversi jawaban teks ke skala numerik ordinal
freq_map = {
    'Often': 4,
    'Sometimes': 3,
    'Rarely': 2,
    'Never, but open to it': 1,
    'Never, and not open to it': 0,
    'No answer': np.nan
}

df_clean = df.copy()
for col in q7_cols:
    df_clean[col] = df_clean[col].map(freq_map)

# 3. Buang responden yang memiliki nilai kosong pada kolom q0007
df_clean = df_clean.dropna(subset=q7_cols).reset_index(drop=True)

print(f"Jumlah responden awal    : {df.shape[0]}")
print(f"Jumlah responden bersih  : {df_clean.shape[0]}")
print(f"Total missing di q0007   : {df_clean[q7_cols].isna().sum().sum()}")
df_clean[q7_cols].describe().round(2)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

q7_labels = {
    'q0007_0001': 'Ask friend: professional advice',
    'q0007_0002': 'Ask friend: personal advice',
    'q0007_0003': 'Physical affection to male friends',
    'q0007_0004': 'Cry',
    'q0007_0005': 'Get in a physical fight',
    'q0007_0006': 'Sexual relations with women',
    'q0007_0007': 'Sexual relations with men',
    'q0007_0008': 'Watch sports',
    'q0007_0009': 'Work out',
    'q0007_0010': 'See a therapist',
    'q0007_0011': 'Feel lonely or isolated'
}

df_q7_labeled = df_clean[q7_cols].rename(columns=q7_labels)

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Panel 1: Rata-rata skor frekuensi (0-4) tiap perilaku
means = df_q7_labeled.mean().sort_values(ascending=True)
axes[0].barh(means.index, means.values, color='#3498db')
axes[0].set_title('Rata-Rata Frekuensi Perilaku (Skala 0–4)', fontsize=12, pad=10)
axes[0].set_xlabel('Skor Rata-Rata (0 = Never & not open, 4 = Often)')
axes[0].set_xlim(0, 4)
axes[0].grid(True, axis='x', linestyle=':', alpha=0.5)

# Panel 2: Heatmap korelasi antar perilaku
corr_matrix = df_q7_labeled.corr()
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    vmin=-0.5,
    vmax=0.8,
    cbar_kws={'shrink': 0.8},
    ax=axes[1]
)
axes[1].set_title('Korelasi Antar Perilaku Survei (q0007)', fontsize=12, pad=10)

plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# 1. Pilih 6 fitur perilaku keterbukaan emosional dan dukungan sosial
selected_features = [
    'q0007_0001',  # Minta saran profesional ke teman
    'q0007_0002',  # Minta saran pribadi ke teman
    'q0007_0003',  # Afeksi fisik ke teman pria
    'q0007_0004',  # Menangis
    'q0007_0010',  # Konsultasi ke terapis
    'q0007_0011'   # Merasa kesepian atau terisolasi
]

X_selected = df_clean[selected_features].copy()

# 2. Samain skala fitur biar bobot jarak K-Means adil
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_selected)

# 3. Uji jumlah klaster k = 2 sampai 8
k_range = range(2, 9)
inertias = []
sil_scores = []

for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    inertias.append(km.inertia_)
    sil_scores.append(silhouette_score(X_scaled, labels))
    print(f"k = {k} | Inertia: {km.inertia_:7.2f} | Silhouette Score: {sil_scores[-1]:.4f}")

# 4. Visualisasi kurva Elbow & Silhouette Score
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

axes[0].plot(k_range, inertias, marker='o', color='#2b5c8f', linewidth=2)
axes[0].set_title('Elbow Method (Inertia per k)')
axes[0].set_xlabel('Jumlah Klaster (k)')
axes[0].set_ylabel('Inertia')
axes[0].grid(True, linestyle=':', alpha=0.6)

axes[1].plot(k_range, sil_scores, marker='o', color='#2ecc71', linewidth=2)
axes[1].set_title('Silhouette Score per k')
axes[1].set_xlabel('Jumlah Klaster (k)')
axes[1].set_ylabel('Silhouette Score')
axes[1].grid(True, linestyle=':', alpha=0.6)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.cluster import KMeans

# 1. Latih model K-Means final dengan k = 2
kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
df_clean['cluster'] = kmeans.fit_predict(X_scaled)

# 2. Cek jumlah dan persentase responden di tiap klaster
cluster_counts = df_clean['cluster'].value_counts().sort_index()
cluster_pct = (cluster_counts / len(df_clean) * 100).round(2)

print("Distribusi Responden per Klaster:")
for cid in cluster_counts.index:
    print(f"- Klaster {cid}: {cluster_counts[cid]} orang ({cluster_pct[cid]}%)")

# 3. Bandingin rata-rata skor asli (skala 0-4) tiap fitur di kedua klaster
feature_names = {
    'q0007_0001': 'Ask friend: professional advice',
    'q0007_0002': 'Ask friend: personal advice',
    'q0007_0003': 'Physical affection to male friends',
    'q0007_0004': 'Cry',
    'q0007_0010': 'See a therapist',
    'q0007_0011': 'Feel lonely or isolated'
}

cluster_profile = (
    df_clean.groupby('cluster')[selected_features]
    .mean()
    .rename(columns=feature_names)
    .T
    .round(2)
)
cluster_profile.columns = ['Klaster 0 (Mean)', 'Klaster 1 (Mean)']
cluster_profile['Selisih (K0 - K1)'] = (cluster_profile['Klaster 0 (Mean)'] - cluster_profile['Klaster 1 (Mean)']).round(2)

print("\nPerbandingan Skor Rata-Rata Perilaku (Skala 0–4):")
print(cluster_profile)
```

```python
from pathlib import Path
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import pandas as pd

# 1. Bandingin persepsi maskulinitas & demografi antar klaster (dalam %)
profile_cols = {
    'q0001': 'Seberapa maskulin responden merasa (q0001)',
    'q0002': 'Seberapa penting dianggap maskulin (q0002)',
    'age3': 'Kelompok Umur',
    'educ4': 'Tingkat Pendidikan',
    'kids': 'Status Memiliki Anak'
}

for col, title in profile_cols.items():
    if col in df_clean.columns:
        print(f"\n=== {title} (%) ===")
        ct = pd.crosstab(df_clean['cluster'], df_clean[col], normalize='index') * 100
        print(ct.round(1))

# 2. Buat folder assets/ dan simpan visualisasi portofolio (2 panel)
assets_dir = Path('../assets')
assets_dir.mkdir(parents=True, exist_ok=True)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Panel 1: Perbandingan Skor Rata-Rata Perilaku per Klaster
plot_df = cluster_profile[['Klaster 0 (Mean)', 'Klaster 1 (Mean)']].iloc[::-1]
y_pos = range(len(plot_df))
bar_h = 0.36

axes[0].barh([y + bar_h/2 for y in y_pos], plot_df['Klaster 0 (Mean)'], height=bar_h,
             color='#2ecc71', label='Klaster 0: Emotionally Open (64.3%)')
axes[0].barh([y - bar_h/2 for y in y_pos], plot_df['Klaster 1 (Mean)'], height=bar_h,
             color='#e67e22', label='Klaster 1: Emotionally Reserved (35.7%)')
axes[0].set_yticks(list(y_pos))
axes[0].set_yticklabels(plot_df.index)
axes[0].set_xlim(0, 3.6)
axes[0].set_xlabel('Skor Rata-Rata Frekuensi (Skala 0–4)')
axes[0].set_title('Profil Perilaku Keterbukaan Emosional per Klaster', fontsize=12, pad=10)
axes[0].legend(loc='lower right')
axes[0].grid(True, axis='x', linestyle=':', alpha=0.5)

# Panel 2: Proyeksi 2D PCA dari 6 Fitur Terpilih
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
var_exp = pca.explained_variance_ratio_ * 100

for cid, color, label in [
    (0, '#2ecc71', 'Klaster 0: Emotionally Open'),
    (1, '#e67e22', 'Klaster 1: Emotionally Reserved')
]:
    mask = df_clean['cluster'] == cid
    axes[1].scatter(X_pca[mask, 0], X_pca[mask, 1], c=color, label=label, alpha=0.65, edgecolors='none', s=28)

axes[1].set_title(f'Visualisasi Klaster K-Means (k = 2) via PCA\n(Total Variansi Dijelaskan: {var_exp.sum():.1f}%)', fontsize=12, pad=10)
axes[1].set_xlabel(f'PC1 ({var_exp[0]:.1f}% variansi)')
axes[1].set_ylabel(f'PC2 ({var_exp[1]:.1f}% variansi)')
axes[1].legend()
axes[1].grid(True, linestyle=':', alpha=0.5)

plt.tight_layout()
plt.savefig(assets_dir / 'cluster_analysis.png', dpi=300, bbox_inches='tight')
plt.show()
print("\nGrafik tersimpan di:", (assets_dir / 'cluster_analysis.png').resolve())
```
